```python
import pandas as pd

# 1. Ler os arquivos
vendas = pd.read_csv("vendas (1).csv")
recuperacoes = pd.read_csv("recuperacoes.csv")

# 2. Agrupar as recuperações por venda
resumo_recuperacoes = (
    recuperacoes
    .groupby("id_venda")
    .agg(
        total_tentativas=("id_recuperacao", "count"),
        valor_recuperado=("valor_recuperado", "sum")
    )
    .reset_index()
)

# 3. Cruzar as bases
base_consolidada = vendas.merge(
    resumo_recuperacoes,
    on="id_venda",
    how="left"
)

# 4. Tratar vendas sem tentativas de recuperação
base_consolidada["total_tentativas"] = (
    base_consolidada["total_tentativas"].fillna(0)
)

base_consolidada["valor_recuperado"] = (
    base_consolidada["valor_recuperado"].fillna(0)
)

# 5. Identificar recuperações sem venda correspondente
recuperacoes_sem_venda = recuperacoes[
    ~recuperacoes["id_venda"].isin(vendas["id_venda"])
]

# 6. Exibir resultados
print("BASE CONSOLIDADA:")
display(base_consolidada)

print("RECUPERAÇÕES SEM VENDA CORRESPONDENTE:")
display(recuperacoes_sem_venda)

print("Dimensão da base consolidada:", base_consolidada.shape)
print("Registros sem correspondência:", len(recuperacoes_sem_venda))
```

```python

# Indicadores de vendas
vendas_aprovadas = vendas[
    vendas["status_pagamento"] == "aprovado"
]

total_vendas_aprovadas = len(vendas_aprovadas)
valor_vendas_aprovadas = vendas_aprovadas["valor_venda"].sum()

ticket_medio = (
    valor_vendas_aprovadas / total_vendas_aprovadas
    if total_vendas_aprovadas > 0 else 0
)

# Indicadores de recuperação
vendas_elegiveis = base_consolidada[
    base_consolidada["elegivel_recuperacao"] == True
]

vendas_recuperadas = vendas_elegiveis[
    vendas_elegiveis["valor_recuperado"] > 0
]

quantidade_elegiveis = len(vendas_elegiveis)
quantidade_recuperadas = len(vendas_recuperadas)

valor_elegivel = vendas_elegiveis["valor_venda"].sum()
valor_recuperado = vendas_recuperadas["valor_recuperado"].sum()

taxa_recuperacao = (
    quantidade_recuperadas / quantidade_elegiveis * 100
    if quantidade_elegiveis > 0 else 0
)

taxa_recuperacao_valor = (
    valor_recuperado / valor_elegivel * 100
    if valor_elegivel > 0 else 0
)

# Exibir indicadores
indicadores = pd.DataFrame({
    "Indicador": [
        "Vendas aprovadas",
        "Valor das vendas aprovadas",
        "Ticket médio",
        "Vendas elegíveis para recuperação",
        "Vendas recuperadas",
        "Valor recuperado",
        "Taxa de recuperação por quantidade",
        "Taxa de recuperação por valor"
    ],
    "Resultado": [
        total_vendas_aprovadas,
        round(valor_vendas_aprovadas, 2),
        round(ticket_medio, 2),
        quantidade_elegiveis,
        quantidade_recuperadas,
        round(valor_recuperado, 2),
        round(taxa_recuperacao, 2),
        round(taxa_recuperacao_valor, 2)
    ]
})

display(indicadores)
```

```python

alertas = []

# 1. Recuperações sem venda correspondente
for _, linha in recuperacoes_sem_venda.iterrows():
    alertas.append({
        "Tipo": "Qualidade dos dados",
        "ID": linha["id_venda"],
        "Alerta": "Recuperação vinculada a uma venda inexistente",
        "Prioridade": "Alta"
    })

# 2. Vendas elegíveis que ainda não foram recuperadas
vendas_nao_recuperadas = base_consolidada[
    (base_consolidada["elegivel_recuperacao"] == True) &
    (base_consolidada["valor_recuperado"] == 0)
]

for _, linha in vendas_nao_recuperadas.iterrows():
    alertas.append({
        "Tipo": "Acompanhamento operacional",
        "ID": linha["id_venda"],
        "Alerta": "Venda elegível sem recuperação bem-sucedida",
        "Prioridade": "Média"
    })

# 3. Exibir painel de alertas
painel_alertas = pd.DataFrame(alertas)

display(painel_alertas)

print("Total de alertas:", len(painel_alertas))
```

```python

import matplotlib.pyplot as plt

# Criar painel visual
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Gráfico 1: indicadores de recuperação
categorias = ["Elegíveis", "Recuperadas", "Não recuperadas"]
valores = [
    quantidade_elegiveis,
    quantidade_recuperadas,
    quantidade_elegiveis - quantidade_recuperadas
]

axes[0].bar(categorias, valores)
axes[0].set_title("Acompanhamento das recuperações")
axes[0].set_ylabel("Quantidade de vendas")

for i, valor in enumerate(valores):
    axes[0].text(i, valor + 0.1, str(valor), ha="center")

# Gráfico 2: alertas
tipos = painel_alertas["Tipo"].value_counts()

axes[1].bar(tipos.index, tipos.values)
axes[1].set_title("Alertas identificados")
axes[1].set_ylabel("Quantidade")
axes[1].tick_params(axis="x", rotation=15)

for i, valor in enumerate(tipos.values):
    axes[1].text(i, valor + 0.1, str(valor), ha="center")

plt.suptitle("YouShop Data Intelligence", fontsize=16)
plt.tight_layout()
plt.show()

# Exibir indicadores e alertas detalhados
print("INDICADORES")
display(indicadores)

print("ALERTAS")
display(painel_alertas)
```
